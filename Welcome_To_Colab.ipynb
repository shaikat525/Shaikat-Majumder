```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import warnings
warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────
# 1.  CONFIGURATION
# ─────────────────────────────────────────────
CSV_PATH = "lstm_ready_voltage_frequency_timeseries_from_homer.csv"

# Input features (exogenous + endogenous)
FEATURE_COLS = [
    "solar_irradiance_W_m2", "wind_speed_m_s", "load_demand_kW",
    "temperature_C", "wind_direction_deg", "humidity_pct", "pressure_hPa",
    "pv_power_kW", "wind_power_kW", "renewable_power_kW",
    "battery_soc_pct", "battery_power_kW_positive_discharge",
    "grid_power_kW_positive_import", "grid_power_kW_positive_export",
    "active_power_P_kW", "reactive_power_Q_kVAr",
    "power_imbalance_kW_positive_deficit",
    "pcc_voltage_pu", "frequency_Hz",           # current values as features
]

# Forecast horizons and their column names in the CSV
HORIZONS = {
    "15min":  ("pcc_voltage_pu_tplus_15min",  "frequency_Hz_tplus_15min"),
    "30min":  ("pcc_voltage_pu_tplus_30min",  "frequency_Hz_tplus_30min"),
    "1h":     ("pcc_voltage_pu_tplus_1h",     "frequency_Hz_tplus_1h"),
    "3h":     ("pcc_voltage_pu_tplus_3h",     "frequency_Hz_tplus_3h"),
    "6h":     ("pcc_voltage_pu_tplus_6h",     "frequency_Hz_tplus_6h"),
    "24h":    ("pcc_voltage_pu_tplus_24h",    "frequency_Hz_tplus_24h"),
}

# Target for this run (change to any key in HORIZONS)
TARGET_HORIZON = "1h"

LOOKBACK       = 16       # past steps fed to LSTM (16 × 15 min = 4 h)
BATCH_SIZE     = 32
EPOCHS         = 50
LR             = 1e-3
HIDDEN_SIZE    = 128
NUM_LAYERS     = 2
DROPOUT        = 0.2
TRAIN_RATIO    = 0.70
VAL_RATIO      = 0.15
SEED           = 42

torch.manual_seed(SEED)
np.random.seed(SEED)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {DEVICE}")


# ─────────────────────────────────────────────
# 2.  LOAD & PREPROCESS
# ─────────────────────────────────────────────
def load_data(path: str, horizon: str):
    df = pd.read_csv(path, parse_dates=["timestamp"])
    df.sort_values("timestamp", inplace=True)
    df.reset_index(drop=True, inplace=True)

    volt_col, freq_col = HORIZONS[horizon]
    target_cols = [volt_col, freq_col]

    # Drop rows where any target is NaN (tail rows with no future label)
    valid = df[target_cols].notna().all(axis=1)
    df = df[valid].reset_index(drop=True)

    X = df[FEATURE_COLS].values.astype(np.float32)
    y = df[target_cols].values.astype(np.float32)
    timestamps = df["timestamp"]

    return X, y, timestamps, target_cols


X_raw, y_raw, timestamps, target_cols = load_data(CSV_PATH, TARGET_HORIZON)
print(f"Samples after NaN drop: {len(X_raw)}")
print(f"Forecasting horizon : {TARGET_HORIZON}")
print(f"Targets             : {target_cols}")


# ─────────────────────────────────────────────
# 3.  TRAIN / VAL / TEST SPLIT (chronological)
# ─────────────────────────────────────────────
n      = len(X_raw)
n_tr   = int(n * TRAIN_RATIO)
n_val  = int(n * VAL_RATIO)
n_te   = n - n_tr - n_val

X_tr,  X_val,  X_te  = X_raw[:n_tr],  X_raw[n_tr:n_tr+n_val],  X_raw[n_tr+n_val:]
y_tr,  y_val,  y_te  = y_raw[:n_tr],  y_raw[n_tr:n_tr+n_val],  y_raw[n_tr+n_val:]
ts_te = timestamps.iloc[n_tr+n_val:].reset_index(drop=True)

print(f"Train/Val/Test sizes: {n_tr} / {n_val} / {n_te}")


# ─────────────────────────────────────────────
# 4.  SCALING
# ─────────────────────────────────────────────
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

X_tr_s  = scaler_X.fit_transform(X_tr)
X_val_s = scaler_X.transform(X_val)
X_te_s  = scaler_X.transform(X_te)

y_tr_s  = scaler_y.fit_transform(y_tr)
y_val_s = scaler_y.transform(y_val)
y_te_s  = scaler_y.transform(y_te)


# ─────────────────────────────────────────────
# 5.  SEQUENCE DATASET
# ─────────────────────────────────────────────
class SeqDataset(Dataset):
    def __init__(self, X, y, lookback):
        self.X, self.y, self.lb = X, y, lookback

    def __len__(self):
        return len(self.X) - self.lb

    def __getitem__(self, i):
        seq  = torch.tensor(self.X[i : i + self.lb], dtype=torch.float32)
        tgt  = torch.tensor(self.y[i + self.lb],     dtype=torch.float32)
        return seq, tgt


train_ds = SeqDataset(X_tr_s,  y_tr_s,  LOOKBACK)
val_ds   = SeqDataset(X_val_s, y_val_s, LOOKBACK)
test_ds  = SeqDataset(X_te_s,  y_te_s,  LOOKBACK)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
val_dl   = DataLoader(val_ds,   batch_size=BATCH_SIZE)
test_dl  = DataLoader(test_ds,  batch_size=BATCH_SIZE)


# ─────────────────────────────────────────────
# 6.  LSTM MODEL
# ─────────────────────────────────────────────
class LSTMForecaster(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, output_size, dropout):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size, hidden_size, num_layers,
            batch_first=True, dropout=dropout if num_layers > 1 else 0.0
        )
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Sequential(
            nn.Linear(hidden_size, 64),
            nn.ReLU(),
            nn.Linear(64, output_size),
        )

    def forward(self, x):
        out, _ = self.lstm(x)
        out = self.dropout(out[:, -1, :])   # last timestep
        return self.fc(out)


model = LSTMForecaster(
    input_size  = len(FEATURE_COLS),
    hidden_size = HIDDEN_SIZE,
    num_layers  = NUM_LAYERS,
    output_size = 2,          # voltage + frequency
    dropout     = DROPOUT,
).to(DEVICE)

print(f"\nModel parameters: {sum(p.numel() for p in model.parameters()):,}")


# ─────────────────────────────────────────────
# 7.  TRAINING LOOP
# ─────────────────────────────────────────────
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, patience=5, factor=0.5
)
criterion = nn.MSELoss()

history = {"train_loss": [], "val_loss": []}
best_val, best_state = float("inf"), None

for epoch in range(1, EPOCHS + 1):
    # ── train ──
    model.train()
    tr_loss = 0.0
    for Xb, yb in train_dl:
        Xb, yb = Xb.to(DEVICE), yb.to(DEVICE)
        optimizer.zero_grad()
        pred = model(Xb)
        loss = criterion(pred, yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        tr_loss += loss.item() * len(Xb)
    tr_loss /= len(train_ds)

    # ── validate ──
    model.eval()
    va_loss = 0.0
    with torch.no_grad():
        for Xb, yb in val_dl:
            Xb, yb = Xb.to(DEVICE), yb.to(DEVICE)
            va_loss += criterion(model(Xb), yb).item() * len(Xb)
    va_loss /= len(val_ds)

    scheduler.step(va_loss)
    history["train_loss"].append(tr_loss)
    history["val_loss"].append(va_loss)

    if va_loss < best_val:
        best_val = va_loss
        best_state = {k: v.clone() for k, v in model.state_dict().items()}

    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{EPOCHS}  train={tr_loss:.6f}  val={va_loss:.6f}")

model.load_state_dict(best_state)
print(f"\nBest val loss: {best_val:.6f}")


# ─────────────────────────────────────────────
# 8.  EVALUATION
# ─────────────────────────────────────────────
def predict_all(loader, model, scaler_y):
    model.eval()
    preds, actuals = [], []
    with torch.no_grad():
        for Xb, yb in loader:
            preds.append(model(Xb.to(DEVICE)).cpu().numpy())
            actuals.append(yb.numpy())
    preds   = scaler_y.inverse_transform(np.vstack(preds))
    actuals = scaler_y.inverse_transform(np.vstack(actuals))
    return preds, actuals


preds, actuals = predict_all(test_dl, model, scaler_y)

labels = ["Voltage (pu)", f"Frequency (Hz)"]
print(f"\n{'Metric':<25} {'Voltage':>12} {'Frequency':>12}")
print("-" * 52)
for i, lbl in enumerate(labels):
    mae  = mean_absolute_error(actuals[:, i], preds[:, i])
    rmse = np.sqrt(mean_squared_error(actuals[:, i], preds[:, i]))
    mape = np.mean(np.abs((actuals[:, i] - preds[:, i]) / (actuals[:, i] + 1e-8))) * 100
    print(f"  MAE  {lbl:<18} {mae:>12.5f}")
    print(f"  RMSE {lbl:<18} {rmse:>12.5f}")
    print(f"  MAPE {lbl:<18} {mape:>11.2f}%")
    print()


# ─────────────────────────────────────────────
# 9.  PLOTS
# ─────────────────────────────────────────────
fig, axes = plt.subplots(3, 1, figsize=(14, 12))

# (a) Loss curves
ax = axes[0]
ax.plot(history["train_loss"], label="Train Loss")
ax.plot(history["val_loss"],   label="Val Loss")
ax.set_title("Training & Validation Loss")
ax.set_xlabel("Epoch"); ax.set_ylabel("MSE Loss")
ax.legend(); ax.grid(True, alpha=0.3)

# (b) Voltage forecast
ax = axes[1]
n_plot = min(200, len(actuals))
ax.plot(actuals[:n_plot, 0], label="Actual Voltage",    linewidth=1.5)
ax.plot(preds[:n_plot,   0], label="Predicted Voltage", linewidth=1.5, linestyle="--")
ax.set_title(f"PCC Voltage Forecast — horizon {TARGET_HORIZON} (first {n_plot} test steps)")
ax.set_xlabel("Step"); ax.set_ylabel("Voltage (pu)")
ax.legend(); ax.grid(True, alpha=0.3)

# (c) Frequency forecast
ax = axes[2]
ax.plot(actuals[:n_plot, 1], label="Actual Frequency",    linewidth=1.5)
ax.plot(preds[:n_plot,   1], label="Predicted Frequency", linewidth=1.5, linestyle="--")
ax.set_title(f"Frequency Forecast — horizon {TARGET_HORIZON} (first {n_plot} test steps)")
ax.set_xlabel("Step"); ax.set_ylabel("Frequency (Hz)")
ax.legend(); ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("lstm_forecast_results.png", dpi=150)
plt.show()
print("Plot saved → lstm_forecast_results.png")


# ─────────────────────────────────────────────
# 10.  SAVE MODEL
# ─────────────────────────────────────────────
torch.save({
    "model_state":  best_state,
    "scaler_X":     scaler_X,
    "scaler_y":     scaler_y,
    "feature_cols": FEATURE_COLS,
    "target_cols":  target_cols,
    "horizon":      TARGET_HORIZON,
    "lookback":     LOOKBACK,
    "hidden_size":  HIDDEN_SIZE,
    "num_layers":   NUM_LAYERS,
}, "lstm_voltage_frequency_model.pt")
print("Model saved → lstm_voltage_frequency_model.pt")


# ─────────────────────────────────────────────
# 11.  MULTI-HORIZON QUICK EVALUATION  (optional)
# ─────────────────────────────────────────────
def evaluate_horizon(horizon_key):
    """Train a minimal model on a given horizon and return MAE."""
    Xr, yr, _, _ = load_data(CSV_PATH, horizon_key)
    n    = len(Xr)
    ntr  = int(n * TRAIN_RATIO)
    nval = int(n * VAL_RATIO)

    sx = MinMaxScaler(); sy = MinMaxScaler()
    Xs = sx.fit_transform(Xr)
    ys = sy.fit_transform(yr)

    ds_tr = SeqDataset(Xs[:ntr],          ys[:ntr],          LOOKBACK)
    ds_va = SeqDataset(Xs[ntr:ntr+nval],  ys[ntr:ntr+nval],  LOOKBACK)
    ds_te = SeqDataset(Xs[ntr+nval:],     ys[ntr+nval:],     LOOKBACK)

    dl_tr = DataLoader(ds_tr, batch_size=BATCH_SIZE, shuffle=True)
    dl_te = DataLoader(ds_te, batch_size=BATCH_SIZE)

    m   = LSTMForecaster(len(FEATURE_COLS), 64, 1, 2, 0.1).to(DEVICE)
    opt = torch.optim.Adam(m.parameters(), lr=LR)
    crit = nn.MSELoss()

    for _ in range(20):          # quick 20-epoch run
        m.train()
        for Xb, yb in dl_tr:
            opt.zero_grad()
            crit(m(Xb.to(DEVICE)), yb.to(DEVICE)).backward()
            opt.step()

    pr, ac = predict_all(dl_te, m, sy)
    mae_v = mean_absolute_error(ac[:, 0], pr[:, 0])
    mae_f = mean_absolute_error(ac[:, 1], pr[:, 1])
    return mae_v, mae_f


print("\n── Multi-Horizon MAE Summary (20-epoch quick runs) ──")
print(f"{'Horizon':<8}  {'Voltage MAE':>14}  {'Freq MAE':>12}")
print("-" * 40)
for h in HORIZONS:
    mv, mf = evaluate_horizon(h)
    print(f"{h:<8}  {mv:>14.5f}  {mf:>12.5f}")
```
